```python
from dotenv import load_dotenv
load_dotenv()
```

```python
from langchain_openai import ChatOpenAI

llm = ChatOpenAI(model="gpt-5-nano")

def get_response(prompt: str):
    return llm.invoke(prompt).content
response = get_response("Describe the future of AI in one sentence.")

print(response)
```

```python
from langchain_core.messages import HumanMessage, SystemMessage

messages = [
    SystemMessage(content="You are a helpful assistant that translates English to Telugu."),
    HumanMessage(content="Translate the following English text to Telugu: 'Hello, how are you?'")
]

llm.invoke(messages)

```

```python
from langchain_core.prompts import PromptTemplate
from langchain_core.output_parsers import StrOutputParser

template = "Translate the following English text to Telugu: '{text}'"
prompt = PromptTemplate.from_template(template)

chain = prompt | llm | StrOutputParser()

chain.invoke({"text": "What is your name?"})
```

```python
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser

chat_prompt = ChatPromptTemplate.from_messages([
    ("system", "You are a helpful assistant that translates English to Telugu."),
    ("human", "Translate the following English text to Telugu: '{text}'")
])

chain = chat_prompt | llm | StrOutputParser()
chain.invoke({"text": "Where do you live?"})
```

```python
template = "Translate the following English text to Telugu: '{text}'"
prompt = PromptTemplate.from_template(template)

from langchain_core.globals import set_llm_cache
from langchain_core.caches import InMemoryCache
set_llm_cache(InMemoryCache())
chain = prompt | llm | StrOutputParser()
chain.invoke({"text": "What is your name?"})
```
